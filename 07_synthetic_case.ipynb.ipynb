```python
import pandas as pd
import numpy as np

df = pd.read_csv(
    "../data/processed/gps_anomalies.csv"
)

df["timestamp"] = pd.to_datetime(
    df["timestamp"]
)

print("GPS data loaded:", df.shape)
```

```python
valid_data = df[
    df["cluster"] != -1
].copy()

print(
    "Valid records:",
    len(valid_data)
)
```

```python
np.random.seed(42)

cases = []

for i in range(100):

    record = valid_data.sample(
        1,
        random_state=42 + i
    ).iloc[0]

    case_time = pd.to_datetime(
        record["timestamp"]
    )

    case = {
        "Case_ID": f"MP-2026-{i+1:03d}",
        "Person_ID": f"P{i+1:03d}",

        "Age_Group": np.random.choice([
            "18-25",
            "26-35",
            "36-45",
            "46-60"
        ]),

        "Gender": np.random.choice([
            "Male",
            "Female",
            "Not Specified"
        ]),

        "Last_Latitude":
            record["latitude"],

        "Last_Longitude":
            record["longitude"],

        "Last_Seen_Time":
            case_time,

        "Day":
            case_time.day_name(),

        "Weather": np.random.choice([
            "Clear",
            "Cloudy",
            "Rain"
        ]),

        "Usual_Area":
            record["cluster"],

        "Average_Distance":
            record["distance_km"],

        "Average_Speed":
            record["speed_kmh"],

        "Previous_Area":
            record["cluster"],

        "Time_Since_Last_Seen":
            np.random.choice([
                1, 2, 3, 4, 6, 8
            ]),

        "Target_Area":
            record["cluster"]
    }

    cases.append(case)
```

```python
synthetic_cases = pd.DataFrame(
    cases
)

print(
    "Synthetic dataset shape:",
    synthetic_cases.shape
)

synthetic_cases.head()
```

```python
import os

os.makedirs(
    "../data/synthetic",
    exist_ok=True
)
```

```python
synthetic_cases.to_csv(
    "../data/synthetic/missing_person_cases.csv",
    index=False
)

print(
    "Synthetic case dataset saved successfully!"
)
```
